```python
caminhos_candidatos = [
    Path(
        "/kaggle/input/datasets/"
        "paultimothymooney/chest-xray-pneumonia/"
        "chest_xray"
    ),
    Path(
        "/kaggle/input/datasets/"
        "paultimothymooney/chest-xray-pneumonia/"
        "chest_xray/chest_xray"
    ),
    Path(
        "/kaggle/input/datasets/"
        "paultimothymooney/chest-xray-pneumonia/"
        "chest_xray/__MACOSX/chest_xray"
    ),
]

divisoes = ["train", "val", "test"]
classes = ["NORMAL", "PNEUMONIA"]
extensoes_validas = {".jpg", ".jpeg", ".png"}
```

```python
comparacao_caminhos = []

for caminho_base in caminhos_candidatos:
    total_caminho = 0

    for divisao in divisoes:
        for classe in classes:
            pasta_classe = caminho_base / divisao / classe

            if pasta_classe.exists():
                arquivos = [
                    arquivo
                    for arquivo in pasta_classe.iterdir()
                    if arquivo.is_file()
                    and arquivo.suffix.lower() in extensoes_validas
                    and not arquivo.name.startswith("._")
                ]

                quantidade = len(arquivos)
            else:
                quantidade = 0

            total_caminho += quantidade

            comparacao_caminhos.append(
                {
                    "caminho": str(caminho_base),
                    "divisao": divisao,
                    "classe": classe,
                    "quantidade": quantidade,
                }
            )

    print("=" * 80)
    print("Caminho:", caminho_base)
    print("Total de imagens válidas:", total_caminho)
```

```python
df_comparacao = pd.DataFrame(comparacao_caminhos)

tabela_comparacao = df_comparacao.pivot_table(
    index="caminho",
    columns=["divisao", "classe"],
    values="quantidade",
    fill_value=0,
)

tabela_comparacao
```

```python
registros = []

for divisao in divisoes:
    for classe in classes:
        pasta_classe = base_dir / divisao / classe

        for arquivo in sorted(pasta_classe.iterdir()):
            arquivo_valido = (
                arquivo.is_file()
                and arquivo.suffix.lower() in extensoes_validas
                and not arquivo.name.startswith("._")
            )

            if arquivo_valido:
                registros.append(
                    {
                        "caminho": str(arquivo),
                        "nome_arquivo": arquivo.name,
                        "divisao_original": divisao,
                        "classe": classe,
                        "extensao": arquivo.suffix.lower(),
                    }
                )

inventario = pd.DataFrame(registros)

print("Quantidade de imagens registradas:")
print(len(inventario))

inventario.head()
```

```python
resumo_divisao_classe = (
    inventario
    .groupby(["divisao_original", "classe"])
    .size()
    .unstack(fill_value=0)
)

resumo_divisao_classe["TOTAL"] = resumo_divisao_classe.sum(axis=1)

resumo_divisao_classe
```

```python
def calcular_sha256(caminho_arquivo):
    sha256 = hashlib.sha256()

    with open(caminho_arquivo, "rb") as arquivo:
        for bloco in iter(lambda: arquivo.read(8192), b""):
            sha256.update(bloco)

    return sha256.hexdigest()
```

```python
from sklearn.model_selection import StratifiedGroupKFold

print("StratifiedGroupKFold importado com sucesso.")
```

```python
base_modelagem_folds = (
    base_modelagem
    .reset_index(drop=True)
    .copy()
)

print("Tamanho:", base_modelagem_folds.shape)

print(
    "Primeiro índice:",
    base_modelagem_folds.index.min()
)

print(
    "Último índice:",
    base_modelagem_folds.index.max()
)
```

```python
sgkf = StratifiedGroupKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

print(sgkf)
```

```python
base_modelagem_folds["fold"] = -1

X = base_modelagem_folds["caminho"]
y = base_modelagem_folds["classe"]
groups = base_modelagem_folds["grupo_candidato"]

for numero_fold, (_, indices_fold) in enumerate(
    sgkf.split(X, y, groups)
):
    base_modelagem_folds.loc[
        indices_fold,
        "fold"
    ] = numero_fold
```

```python
print(
    "Registros sem fold:",
    (base_modelagem_folds["fold"] == -1).sum()
)

print("\nQuantidade de imagens por fold:")

print(
    base_modelagem_folds["fold"]
    .value_counts()
    .sort_index()
)
```

```python
folds_por_grupo = (
    base_modelagem_folds
    .groupby("grupo_candidato")["fold"]
    .nunique()
)

print(
    "Maior quantidade de folds para um mesmo grupo:",
    folds_por_grupo.max()
)

print(
    "Grupos presentes em mais de um fold:",
    (folds_por_grupo > 1).sum()
)
```

```python
classes_por_fold = (
    base_modelagem_folds
    .groupby(["fold", "classe"])
    .size()
    .unstack(fill_value=0)
)

print(classes_por_fold)
```

```python
percentuais_por_fold = (
    classes_por_fold
    .div(classes_por_fold.sum(axis=1), axis=0)
    * 100
).round(2)

print(percentuais_por_fold)
```

```python
percentual_normal_global = (
    (base_modelagem_folds["classe"] == "NORMAL").mean()
    * 100
)

percentual_pneumonia_global = (
    (base_modelagem_folds["classe"] == "PNEUMONIA").mean()
    * 100
)

print(
    "NORMAL global:",
    round(percentual_normal_global, 2),
    "%"
)

print(
    "PNEUMONIA global:",
    round(percentual_pneumonia_global, 2),
    "%"
)
```

```python
desvio_normal = (
    percentuais_por_fold["NORMAL"]
    - percentual_normal_global
).abs()

print("Desvio absoluto da proporção NORMAL por fold:")
print(desvio_normal.round(2))

print(
    "\nMaior desvio encontrado:",
    round(desvio_normal.max(), 2),
    "pontos percentuais"
)
```

```python
base_modelagem_folds["divisao_nova"] = "treino"

base_modelagem_folds.loc[
    base_modelagem_folds["fold"] == 0,
    "divisao_nova"
] = "teste"

base_modelagem_folds.loc[
    base_modelagem_folds["fold"] == 1,
    "divisao_nova"
] = "validacao"
```

```python
contagem_divisao_nova = (
    base_modelagem_folds["divisao_nova"]
    .value_counts()
)

print(contagem_divisao_nova)

print("\nPercentuais:")

print(
    (
        base_modelagem_folds["divisao_nova"]
        .value_counts(normalize=True)
        * 100
    ).round(2)
)
```

```python
percentuais_classes_divisao = (
    classes_por_divisao_nova
    .div(
        classes_por_divisao_nova.sum(axis=1),
        axis=0
    )
    * 100
).round(2)

print(percentuais_classes_divisao)
```
